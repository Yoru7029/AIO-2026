```python
import torch
import torchvision.models as models
from torchvision.models import VGG16_Weights


# Load the pretrained VGG16 model
model = models.vgg16()
```

```python
model.features
```

```python
# Iterate over all modules in the model
for name, module in model.named_modules():
    # Check if the module has parameters (like Conv2d, Linear, etc.)
    if hasattr(module, 'parameters'):
        # Check if any parameter in the module requires gradient
        is_trainable = any(param.requires_grad for param in module.parameters())
        print(f"{name}: {'Trainable' if is_trainable else 'Not trainable'}")
```

```python
vgg16 = models.vgg16()
f_extractor = vgg16.features

# Freeze the first five layers
layer_count = 0
for child in f_extractor.children():
    if layer_count < 5:
        for param in child.parameters():
            param.requires_grad = False
    layer_count += 1

# Check
for name, param in vgg16.named_parameters():
    print(f"{name}: {param.requires_grad}")
```

```python
import torch.nn as nn
from torchsummary import summary

model = nn.Sequential(f_extractor, nn.Flatten(), nn.Dropout(), nn.Linear(512, 10))
summary(model, (3, 32, 32))
```

```python
import torch.nn as nn
import torch

features = nn.Sequential(nn.Conv2d(1, 6, kernel_size=5),
                       nn.ReLU(),
                       nn.MaxPool2d(2, 2)
)

x = torch.randn(1, 1, 28, 28)
output1 = features(x)
print(f"Output shape: {output1.shape}")
```

```python
model = nn.Sequential(features,
                      nn.Flatten(),
                      nn.Linear(12*12*6, 10)
)

x = torch.randn(1, 1, 28, 28)
output2 = model(x)
print(f"Output shape: {output2.shape}")
```
