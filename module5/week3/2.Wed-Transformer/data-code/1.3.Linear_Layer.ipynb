## Example 1

```python
import torch 
import torch.nn as nn
import torch.nn.functional as F

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 4)
        self.fc2 = nn.Linear(4, 3)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

model = MyModel()
```

```python
input_x = torch.randn((32, 5))
output = model(input_x)
print(output.shape)
```

```python
from torchinfo import  summary

input_x = torch.randn((32, 5))
model = MyModel()

summary(model, input_data=input_x)
```

```python
from torchinfo import  summary

input_x = torch.randn((32, 8, 5))
model = MyModel()

summary(model, input_data=input_x)
```

## Example 2

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(100, 5)
        self.fc2 = nn.Linear(5, 3)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x
```

```python
from torchinfo import  summary

input_x = torch.randn((32, 10, 100))
model = MyModel()

summary(model, input_data=input_x)
```
