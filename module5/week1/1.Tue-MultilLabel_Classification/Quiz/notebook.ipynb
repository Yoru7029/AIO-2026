## Q5

```python
x = np.array([1, 2, 1])
y = np.array([1, 0, 1])

W = np.array([[0, 1, -1],
              [0, 0, 0],
              [0, 0, 0]])

z = x @ W
print(f"\nz: \n{z}")
cond_list = [z == 0, z == 1, z == -1]
val_list = [0.5, 0.731, 0.269]
yhat = np.select(condlist=cond_list, choicelist=val_list, default=np.nan)
print(f"\nyhat: \n{yhat}")

dz = yhat - y
print(f"\ndz: \n{dz}")
grad_w = np.outer(x, dz)
print(f"\ngrad_w: \n{grad_w}")
eta = 0.1
print("\n", np.round(W - eta*grad_w, 3))
```
