# Step 1: Import libraries

```python
import numpy as np
```

# Step 2: Define support functions

## Vector length (L2 norm)

```python
def compute_vector_length(vector: np.ndarray) -> float:
    # YOUR CODE HERE
    # Hint: L2 norm = sqrt(sum(x_i^2))
    return np.sqrt((vector**2).sum())

vector = np.array([-2, 4, 9, 21])
result = compute_vector_length(vector)
print(round(result, 2))
```

```python
np.linalg.norm(vector)
```

## Dot product

```python
def compute_dot_product(vector1: np.ndarray, vector2: np.ndarray) -> float:
    return float(np.dot(vector1, vector2))

v1 = np.array([0, 1, -1, 2])
v2 = np.array([2, 5, 1, 0])
result = compute_dot_product(v1, v2)
print(round(result, 2))
```

# Step 3: Define cosine similarity function

```python
def compute_cosine(v1: np.ndarray, v2: np.ndarray) -> float:
    dot_product = compute_dot_product(v1, v2)
    # YOUR CODE HERE
    # Hint: use compute_vector_length to get norm_v1 and norm_v2
    norm_v1 = compute_vector_length(v1)
    norm_v2 = compute_vector_length(v2)
    return dot_product / (norm_v1 * norm_v2)


x = np.array([1, 2, 3, 4])
y = np.array([1, 0, 3, 0])
result = compute_cosine(x, y)
print(round(result, 3))
```

## Alternative approach: Using a library built-in function

```python
from scipy.spatial.distance import cosine
```

```python
def compute_cosine_scipy(v1: np.ndarray, v2: np.ndarray) -> float:
    # YOUR CODE HERE
    # Hint: scipy.spatial.distance.cosine returns 1 - cosine_similarity
    return 1 - cosine(v1, v2)

result = compute_cosine_scipy(x, y)
print(round(result, 3))
```
