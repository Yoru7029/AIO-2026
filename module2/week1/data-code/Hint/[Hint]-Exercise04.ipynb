# Step 3: Analayze the dataset

```python
data = df.to_numpy()

tv = data[:, 0]
radio = data[:, 1]
newspaper = data[:, 2]
sales = data[:, -1]
```

## Max sales value and its index

```python
# YOUR CODE HERE
max_value = np.max(sales)
max_index = np.argmax(sales)
print(f"Max: {max_value} - Index: {max_index}")
```

## Mean of the TV column

```python
# YOUR CODE HERE
mean_tv = np.mean(tv)
print(mean_tv)
```

## Count records with Sales >= 20

```python
# YOUR CODE HERE
count = np.sum(sales >= 20)
print(count)
```

## Mean Radio spend where Sales is above average

```python
# YOUR CODE HERE
mean_radio = np.mean(radio[sales > sales.mean()])
print(mean_radio)
```

## Total Sales where Newspaper spend is above average

```python
mean_newspaper = np.mean(newspaper)
# YOUR CODE HERE
sum_sales = np.sum(sales[newspaper > mean_newspaper])
print(sum_sales)
```
