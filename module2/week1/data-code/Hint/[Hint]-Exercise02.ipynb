# Step 1: Import libraries

```python
import numpy as np
import matplotlib.image as mpimg
```

```python
# Support function for visualization
import matplotlib.pyplot as plt

def show_image(img: np.ndarray, title: str = "") -> None:
    plt.imshow(img, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.show()
```

## Step 3: Implement grayscale conversion functions

## Method 1: Lightness -> (max(R, G, B) + min(R, G, B)) / 2

```python
def to_gray_lightness(img: np.ndarray) -> np.ndarray:
    # YOUR CODE HERE
    return (img.max(axis=2) + img.min(axis=2))/2


gray_img_01 = to_gray_lightness(img)
print(gray_img_01[0, 0])
```

```python
gray_img_01.shape
```

```python
show_image(gray_img_01, "Lightness")
```

## Method 2: Average -> (R + G + B) / 3

```python
def to_gray_average(img: np.ndarray) -> np.ndarray:
    # YOUR CODE HERE
    return img.mean(axis=2)


gray_img_02 = to_gray_average(img)
print(gray_img_02[0, 0])
```

```python
show_image(gray_img_02, "Average")
```

## Method 3: Luminosity -> 0.21 * R + 0.72 * G + 0.07 * B

```python
def to_gray_luminosity(img: np.ndarray) -> np.ndarray:
    # YOUR CODE HERE
    lumi = np.array([0.21, 0.72, 0.07])
    return img @ lumi


gray_img_03 = to_gray_luminosity(img)
print(gray_img_03[0, 0])
```

```python
show_image(gray_img_03, "Luminosity")
```

## Method 4: Using a library built-in function

```python
import cv2
```

```python
def to_gray_cv2(img: np.ndarray) -> np.ndarray:
    # YOUR CODE HERE
    # Hint: cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    return cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)


gray_img_04 = to_gray_cv2(img)
print(gray_img_04[0, 0])
```

```python
show_image(gray_img_04, "cv2.cvtColor")
```
