```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Flatten(), nn.Linear(784, 10)
)
print(model)
```

```python
# Generating a random tensor
input_tensor = torch.rand(5, 28, 28)

# Feeding the tensor into the model
output = model(input_tensor)
print(output.shape)
```

```python
import torch
import torch.nn as nn

# Model definition
model = nn.Sequential(
    nn.Linear(4, 3),
    nn.Sigmoid(),
    nn.Linear(3, 3),
    nn.Softmax(dim=1)
)

# Generating a random tensor of shape [5, 4]
input_tensor = torch.rand(5, 4)

# Feeding the tensor into the model
output = model(input_tensor)
print(output)
```
