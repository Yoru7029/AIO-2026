```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
```

```python
class SimpleMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(1, 2)
        self.layer2 = nn.Linear(2, 2)
        self.layer3 = nn.Linear(2, 1)

    def forward(self, x):
        output1 = torch.relu(self.layer1(x))
        output2 = torch.relu(self.layer2(output1))
        output3 = self.layer3(output2)
        return output3.squeeze(1)
```

```python
model = SimpleMLP()
model.to(device)

w1 = torch.tensor([[0.4967],
                   [-0.1383]], requires_grad=True)
b1 = torch.tensor([0.6477, 1.5230], requires_grad=True)

w2 = torch.tensor([[-0.2342, -0.2341],
                   [-0.2341, -0.2342]], requires_grad=True)
b2 = torch.tensor([1.5792, 0.7674], requires_grad=True)

w3 = torch.tensor([-0.4695, 0.5426], requires_grad=True)
b3 = torch.tensor([0.0], requires_grad=True)


with torch.no_grad():
    model.layer1.weight.copy_(w1)
    model.layer1.bias.copy_(b1)

    model.layer2.weight.copy_(w2)
    model.layer2.bias.copy_(b2)

    model.layer3.weight.copy_(w3)
    model.layer3.bias.copy_(b3)
```

```python
for param in model.parameters():
    print(f"\n{param}")
```

```python
mean_ = 3; std_ = 2
```
