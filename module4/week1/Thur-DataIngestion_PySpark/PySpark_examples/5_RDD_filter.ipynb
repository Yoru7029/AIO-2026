```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySparkExample.com').getOrCreate()
print(spark)
```

```python
num_data = [1,2,3,4,5]

num_rdd = spark.sparkContext.parallelize(num_data)
```

```python
num_rdd_filter = num_rdd.filter(lambda x: x % 2 == 0)
```

```python
num_rdd_filter.collect()
```
