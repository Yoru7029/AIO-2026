```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySparkExample.com').getOrCreate()
print(spark)
```

```python
text_data = [
    "Hello this is an example of Spark WordCount Example",
    "we are using PySpark",
    "PySpark is a great tool for Big Data Analysis"
]

text_rdd = spark.sparkContext.parallelize(text_data)
```

```python
text_rdd2 = text_rdd.map(lambda x: (x, len(x.split(" "))))
```

```python
text_rdd3 = text_rdd2.map(lambda x: (x[0], x[1], x[1]**2))
```

```python
text_rdd2.collect()
```

```python
text_rdd3.collect()
```
