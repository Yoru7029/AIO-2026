```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySparkExample.com').getOrCreate()
print(spark)
```

```python
df_csv = spark.read.csv("./people.csv",
                        header=True,
                        inferSchema=True)
```

```python
df_csv.printSchema()
df_csv.show(truncate=True)
```

```python
df_csv.createOrReplaceTempView("people") 
```

```python
query = '''SELECT name, sex
FROM people'''

df2 = spark.sql(query)
```

```python
df2.show(truncate=True)
```
